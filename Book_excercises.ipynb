```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("mpg")
```

```python
df.info()
```

### Visualisation of Variables with a scatterplot

### Use a normal scatterplot

Model from more recent years have generally lower displacement and horsepower

```python
sns.scatterplot(data=df,x='displacement',y = 'horsepower',hue = 'model_year');
```

### Use in a normal scatterplot 'style' to map year 

```python
sns.scatterplot(data=df,x='displacement',y = 'horsepower',style = 'model_year')
```

### Usin Marker Size to introduce an extra variable

To and higher displacement and horsepower is associated an heavier weight

```python
sns.scatterplot(data=df,x='displacement',y = 'horsepower',size = 'weight')
```

### Visualisation year using intensity through alpha

```python
sns.scatterplot(data=df,x='displacement',y = 'horsepower',size = 'weight',alpha = 0.5)
```

### Comparing with R seaborn neeither matplotlib can visualise a variable using the intensity

```python
plt.scatter(df['displacement'],df['horsepower'],alpha=df['weight']);
```

### Define the color of the points

```python
sns.scatterplot(data = df,x='displacement',y='horsepower',color='red')
```

# GEOMETRIC OBJECTS
### Track a regression line

```python
sns.regplot(data=df,x='displacement',y='horsepower',order = 3)
```

# Regplot doesn't support hue

```python
# add more variables

sns.regplot(data=df,x='displacement',y='horsepower', hue = 'model_year',order = 3)
```

### We can use lmplot 

By default it will create a regression per each year

```python
sns.lmplot(data = df,x='displacement',y='horsepower',hue='model_year')
```

use fit_reg = False to remove multiple regressions

```python
sns.lmplot(data = df,x='displacement',y='horsepower',hue='model_year', fit_reg = False)
```

# We can add the regression by plotting over the lmplot a regressionplot without scatter

```python
sns.lmplot(data = df,x='displacement',y='horsepower',hue='model_year', fit_reg = False)
sns.regplot(data = df,x = 'displacement',y='horsepower',scatter = False, order = 3, color ='r');
```

# can specifiy a class to be highlighted in the chart using filter

```python
sns.scatterplot(data=df, x='displacement', y='horsepower')

sns.scatterplot(
    data=df[(df['displacement'].notna()) &
        (df['horsepower'].notna()) &
        (df['displacement'] > df['displacement'].quantile(0.95)) |
        (df['horsepower'] > df['horsepower'].quantile(0.95))
    ],
    x='displacement',
    y='horsepower',
    color='r'
)
```

# Histplot
